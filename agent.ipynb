```python
def reflex_vacuum_agent(location, status):
    if status == "Dirty":
        return "Suck"
    elif location == "A":
        return "Move Right"
    elif location == "B":
        return "Move Left"


environment={
    "A": "Dirty",
    "B": "Dirty"
}    

location="A"

while True:
    status=environment[location]
    action=reflex_vacuum_agent(location,status)

    print(f"Location :{location} , status :{status} , Action :{action}")

    if action=="Suck":
        environment[location]="Clean"

    elif action=="Move Right":
        location="B"

    elif action=="Move Left":
        location="A"


    if environment["A"]=="Clean" and environment["B"]=="Clean":
        print("Both rooms are clean. Task Completed!")
        break   

    
```

```python
def tower(n, fromrod, torod, auxrod):
    if n > 0:
        tower(n-1, fromrod, auxrod, torod)
        print("Move disk", n, "from rod", fromrod, "to rod", torod)
        tower(n-1, auxrod, torod, fromrod)


n = 3
tower(n, 'A', 'C', 'B')
```

```python
# Simple Rule-Based Chatbot Using Pattern Matching

def chatbot_response(user_input):

    user_input = user_input.lower()

    # Rule 1: Greeting
    if any(word in user_input.split()
           for word in ["hello", "hi", "hey"]):
        return "Hello! How can I help you?"

    # Rule 2: Name
    elif "name" in user_input:
        return "I am a simple AI chatbot."

    # Rule 3: How are you
    elif "how are you" in user_input:
        return "I am fine. Thank you for asking!"

    # Rule 4: Course or Subject
    elif any(word in user_input.split()
             for word in ["course", "subject"]):
        return "I can provide basic information about your course."

    # Rule 5: Help
    elif "help" in user_input:
        return ("Sure! You can ask me about my name, "
                "course, or general information.")

    # Rule 6: Thank You
    elif any(word in user_input.split()
             for word in ["thank", "thanks"]):
        return "You are welcome!"

    # Rule 7: Exit
    elif any(word in user_input.split()
             for word in ["bye", "exit", "quit"]):
        return "Goodbye! Have a nice day."

    # Rule 8: Default Response
    else:
        return "Sorry, I do not understand your question."


# Main Program
print("Chatbot: Hello! Welcome to the chatbot.")
print("Chatbot: Type 'bye' to exit.")

while True:

    user_input = input("You: ")

    response = chatbot_response(user_input)

    print("Chatbot:", response)

    # Stop chatbot after exit command
    if any(word in user_input.lower().split()
           for word in ["bye", "exit", "quit"]):
        break
```
